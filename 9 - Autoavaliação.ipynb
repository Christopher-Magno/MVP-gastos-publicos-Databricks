# Autoavaliação

Acredito que eu tenha atingido o objetivo, mesmo estando totalmente perdido por esse ser meu primeiro contato com a ferramenta. Após muitos vídeos assistidos, muitas perguntas para a Genie e as orientações dos professores nas aulas síncronas, eu consegui fazer o projeto.

E achei incrível o que o Databricks pode fazer. Eu poder por conta própria pegar uma base de dados, de uma fonte confiável, e limpar e estudar sobre ela é muito maneiro. Você sozinho buscar respostas para suas dúvidas através dos dados é muito legal.

Estou adorando aprender sobre ciência de dados e entender o real poder dos dados e aprender a manipulá-los é muito show.

## Dificuldades e Aprendizados

Em relação às etapas do projeto, tive um pouco de dificuldade em saber os comandos para trabalhar com os dados. Escolher os dados e inseri-los eu achei bem tranquilo, a questão é gravar todos os comandos e saber o passo a passo a ser seguido. Para quem não tem experiência, é difícil. Por isso tive que pesquisar muito e recorrer à ajuda, e isso me ajudou a entender bastante sobre o funcionamento do Databricks.

Em relação ao projeto, optei pela arquitetura medalhão, e em cada schema apliquei o tratamento sobre os dados para criar a tabela gold e deixar pronto para aplicação em business intelligence e assim conseguir responder às minhas perguntas.

## Conclusão

De forma geral, adorei esse MVP e aprendi bastante. Esse é meu primeiro módulo e já deu pra ver que vai valer a pena a migraçao de carreira e que vou aprender de verdade nesse curso.
